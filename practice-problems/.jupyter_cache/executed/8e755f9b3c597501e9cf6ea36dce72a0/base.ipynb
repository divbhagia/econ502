```python
#| echo: false
#| fig-align: center

import matplotlib.pyplot as plt
import numpy as np

# Colorblind-friendly palette
CB_BLUE  = '#0072B2'
CB_RED   = '#D55E00'
CB_GREEN = '#009E73'

# ── Parameters ─────────────────────────────────────────────────
px, py0, py1, m = 1, 2, 1, 10

# Key points
xA, yA = 5.0, 2.5                         # original optimum A
xC, yC = 5.0, 5.0                         # new optimum C
u0 = xA * yA                              # = 12.5
xB = np.sqrt(py1 / px * u0)               # Hicksian demand for x at new prices
yB = np.sqrt(px / py1 * u0)               # Hicksian demand for y at new prices  ≈ 3.54
e_comp = px * xB + py1 * yB               # compensated expenditure ≈ 7.07

# ── Figure ─────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 6))

x = np.linspace(0.3, 11, 500)

# Indifference curves
ax.plot(x, u0 / x, color=CB_BLUE, lw=2, label=r'$u(x,y)=12.5$ (original)')
ax.plot(x, (xC * yC) / x, color=CB_GREEN, lw=2, label=r'$u(x,y)=25$ (new)')

# Budget lines
ax.plot(x, (m - px * x) / py0, color=CB_BLUE, lw=1.8, ls='-',
        label=r'$BL_1$: $x + 2y = 10$')
ax.plot(x, (m - px * x) / py1, color=CB_RED, lw=1.8, ls='-',
        label=r'$BL_2$: $x + y = 10$')
ax.plot(x, (e_comp - px * x) / py1, color=CB_RED, lw=1.8, ls='--',
        label=r'$BL_{comp}$: compensated')

# Optimal points
ax.plot(xA, yA, 'o', color=CB_BLUE, ms=8, zorder=5)
ax.plot(xB, yB, 's', color=CB_RED,  ms=8, zorder=5)
ax.plot(xC, yC, 'D', color=CB_GREEN, ms=8, zorder=5)

# Labels for points
ax.annotate(r'$A$  $(5,\;2.5)$', xy=(xA, yA),
            xytext=(xA + 0.3, yA - 0.6), fontsize=12, color=CB_BLUE)
ax.annotate(r'$B$  $(\sqrt{12.5},\;\sqrt{12.5})$', xy=(xB, yB),
            xytext=(xB + 0.3, yB - 0.6), fontsize=12, color=CB_RED)
ax.annotate(r'$C$  $(5,\;5)$', xy=(xC, yC),
            xytext=(xC + 0.3, yC + 0.15), fontsize=12, color=CB_GREEN)

# ── Decomposition arrows on the right margin ────────────────────
x_arr = 10.0  # x-position for arrows
ax.annotate('', xy=(x_arr, yB), xytext=(x_arr, yA),
            arrowprops=dict(arrowstyle='<->', color=CB_RED, lw=1.8))
ax.text(x_arr + 0.25, (yA + yB) / 2, 'SE', va='center', fontsize=12, color=CB_RED)

ax.annotate('', xy=(x_arr, yC), xytext=(x_arr, yB),
            arrowprops=dict(arrowstyle='<->', color=CB_GREEN, lw=1.8))
ax.text(x_arr + 0.25, (yB + yC) / 2, 'IE', va='center', fontsize=12, color=CB_GREEN)

# ── Formatting ──────────────────────────────────────────────────
ax.set_xlim(0, 11)
ax.set_ylim(0, 11)
ax.set_xlabel('Good $x$', fontsize=13)
ax.set_ylabel('Good $y$', fontsize=13)
ax.set_title('Slutsky Decomposition: Substitution and Income Effects', fontsize=13)
ax.legend(fontsize=11, loc='upper right')
ax.spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.show()
```
