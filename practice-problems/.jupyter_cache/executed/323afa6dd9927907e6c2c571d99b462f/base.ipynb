```python
#| echo: false
#| fig-align: center

import matplotlib.pyplot as plt
import numpy as np

# Colorblind-friendly palette
CB_BLUE  = '#0072B2'
CB_RED   = '#D55E00'
CB_GREEN = '#009E73'
CB_ORANGE = '#E69F00'


# ── Parameters ─────────────────────────────────────────────────
px, py0, py1, m = 1, 2, 1, 10

# Key points
xA, yA = 5.0, 2.5                         # original optimum A
xC, yC = 5.0, 5.0                         # new optimum C
u0 = xA * yA                              # = 12.5
xB = np.sqrt(py1 / px * u0)               # Hicksian demand for x at new prices
yB = np.sqrt(px / py1 * u0)               # Hicksian demand for y at new prices  ≈ 3.54
e_comp = px * xB + py1 * yB               # compensated expenditure ≈ 7.07

# ── Figure ─────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(6.5, 5))

x = np.linspace(0.0001, 11, 500)

# Indifference curves
ax.plot(x, u0 / x, color=CB_BLUE, lw=2)
ax.plot(x, (xC * yC) / x, color=CB_RED, lw=2)

# Budget lines
ax.plot(x, (m - px * x) / py0, color=CB_BLUE, lw=1.8, ls='-',
        label=r'$x + 2y = 10$ (original)')
ax.plot(x, (m - px * x) / py1, color=CB_RED, lw=1.8, ls='-',
        label=r'$x + y = 10$ (new)')
ax.plot(x, (e_comp - px * x) / py1, color=CB_BLUE, lw=1.8, ls='--',
        label=r'Compensated budget')

# Optimal points
ax.plot(xA, yA, 'o', color=CB_BLUE, ms=8, zorder=5)
ax.plot(xB, yB, 'o', color=CB_BLUE,  ms=8, zorder=5)
ax.plot(xC, yC, 'o', color=CB_RED, ms=8, zorder=5)

# ── Decomposition arrows on the right margin ────────────────────
x_arr = 10.0  # x-position for arrows
ax.annotate('', xy=(x_arr, yB), xytext=(x_arr, yA),
            arrowprops=dict(arrowstyle='<->', color=CB_ORANGE, lw=1.8))
ax.text(x_arr + 0.25, (yA + yB) / 2, 'SE', va='center', fontsize=14, color=CB_ORANGE)

ax.annotate('', xy=(x_arr, yC), xytext=(x_arr, yB),
            arrowprops=dict(arrowstyle='<->', color=CB_GREEN, lw=1.8))
ax.text(x_arr + 0.25, (yB + yC) / 2, 'IE', va='center', fontsize=14, color=CB_GREEN)

# ── Formatting ──────────────────────────────────────────────────
ax.set_xlim(0, 11)
ax.set_ylim(0, 11)
ax.set_xticks([xA, xB, xC])
ax.set_yticks([yA, yB, yC])
ax.set_xticklabels([f'{xA:.2f}', f'{xB:.2f}', f'{xC:.2f}'], size=12)
ax.set_yticklabels([f'{yA:.2f}', f'{yB:.2f}', f'{yC:.2f}'], size=12)
ax.set_xlabel('Good $x$', fontsize=13)
ax.set_ylabel('Good $y$', fontsize=13)
ax.set_title('Slutsky Decomposition: Substitution and Income Effects', fontsize=13)
ax.legend(fontsize=13, loc='upper right')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#| echo: false
#| fig-align: center
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))

# Parameters
I = 100
T = 100          # transfer amount
px, py = 10, 5
alpha = 0.3      # food share under Cobb-Douglas preferences

# ── Budget lines ──────────────────────────────────────────────
x_orig = np.linspace(0, I/px, 100)
ax.plot(x_orig, (I - px*x_orig)/py, color='gray', lw=2, ls='--',
        alpha=0.7, label='Original budget')

x_cash = np.linspace(0, (I+T)/px, 100)
ax.plot(x_cash, (I+T - px*x_cash)/py, color='black', lw=2.5,
        label=f'Cash transfer (${T})')

x_min = T / px                           # minimum food required by in-kind
x_inkind = np.linspace(x_min, (I+T)/px, 100)
ax.plot(x_inkind, (I - px*(x_inkind - x_min))/py,
        color='#D55E00', lw=9, alpha=0.35,
        label=f'In-kind transfer (${T} food vouchers)')
ax.axvline(x=x_min, color='#D55E00', ls=':', lw=2, alpha=0.6)
ax.text(x_min - 0.3, 11.5, f'Must buy\n≥{x_min:.0f} units\nof food',
        fontsize=9.5, color='#D55E00', ha='right', fontweight='bold')

# ── Optimal points ─────────────────────────────────────────────
x_cash_opt = alpha * (I+T) / px          # cash optimum
y_cash_opt = (1-alpha) * (I+T) / py
x_inkind_opt = x_min                     # in-kind optimum (kink, binding case)
y_inkind_opt = I / py

ax.plot(x_cash_opt, y_cash_opt, 'ko', ms=9, zorder=5, label='Optimum with cash')
ax.plot(x_inkind_opt, y_inkind_opt, 'o', color='#D55E00', ms=9, zorder=5,
        label='Optimum with in-kind')

# ── Indifference curves ────────────────────────────────────────
x_ic = np.linspace(0.5, 30, 400)
U_cash   = x_cash_opt**alpha   * y_cash_opt**(1-alpha)
U_inkind = x_inkind_opt**alpha * y_inkind_opt**(1-alpha)
ax.plot(x_ic, (U_cash   / x_ic**alpha)**(1/(1-alpha)), 'k',  lw=2)
ax.plot(x_ic, (U_inkind / x_ic**alpha)**(1/(1-alpha)), color='#D55E00', lw=2)

# ── Formatting ─────────────────────────────────────────────────
ax.set_xlabel('Food ($x$)', fontsize=13)
ax.set_ylabel('Other goods ($y$)', fontsize=13)
ax.set_xlim(0, 22)
ax.set_ylim(0, 42)
ax.legend(fontsize=11, loc='upper right')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```
