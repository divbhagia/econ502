```python
#| echo: false
#| fig-align: center

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Polygon

# ========== PARAMETERS ==========
# Demand: P = 200 - 2Q  =>  Q = 100 - P/2
# MC = 40 (constant)
a_demand = 200       # P-intercept of demand
b_demand = 2         # slope (P = a - bQ)
MC = 40

Q_m = 40; P_m = 120
Q_c = 80; P_c = 40

# ========== COLORS ==========
CB_BLUE  = '#0072B2'
CB_RED   = '#D55E00'
CB_GREEN = '#009E73'
CB_GRAY  = '#999999'

# ========== FIGURE ==========
fig, ax = plt.subplots(figsize=(7, 5.5))

Q_range = np.linspace(0, 110, 500)
P_demand = a_demand - b_demand * Q_range
MR = a_demand - 2 * b_demand * Q_range

# Curves
mask_d = P_demand >= 0
mask_mr = MR >= -10
ax.plot(Q_range[mask_d], P_demand[mask_d], color=CB_RED, lw=2.5, label='Demand: $P = 200 - 2Q$')
ax.plot(Q_range[mask_mr], MR[mask_mr], color=CB_RED, lw=2, ls='--', label='$MR = 200 - 4Q$')
ax.axhline(MC, color=CB_BLUE, lw=2.5, label='$MC = 40$')

# DWL triangle
verts_dwl = [(Q_m, P_m), (Q_c, P_c), (Q_m, P_c)]
ax.add_patch(Polygon(verts_dwl, closed=True, fc=CB_RED, alpha=0.3, ec='none', label='DWL'))

# Transfer rectangle
verts_transfer = [(0, P_c), (Q_m, P_c), (Q_m, P_m), (0, P_m)]
ax.add_patch(Polygon(verts_transfer, closed=True, fc=CB_BLUE, alpha=0.15, ec='none', label='CS → PS transfer'))

# Dashed guides
ax.plot([Q_m, Q_m], [0, P_m], ':', color=CB_GRAY, lw=0.9)
ax.plot([0, Q_m], [P_m, P_m], ':', color=CB_GRAY, lw=0.9)
ax.plot([Q_c, Q_c], [0, P_c], ':', color=CB_GRAY, lw=0.9)

# Labels
ax.set_xlabel('Quantity ($Q$)', fontsize=12)
ax.set_ylabel('Price ($P$)', fontsize=12)
ax.set_xlim(0, 110)
ax.set_ylim(0, 210)

ax.set_xticks([Q_m, Q_c])
ax.set_xticklabels(['$Q_m = 40$', '$Q_c = 80$'], fontsize=10)
ax.set_yticks([MC, P_m, a_demand])
ax.set_yticklabels(['$MC = 40$', '$P_m = 120$', '200'], fontsize=10)

ax.legend(fontsize=9, loc='upper right', framealpha=0.85)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#| echo: false
#| fig-align: center

import matplotlib.pyplot as plt
import numpy as np

# ========== PARAMETERS ==========
a = 150
b = 1
c = 30

color1 = '#0072B2'
color2 = '#E69F00'
color_eq = '#009E73'
color_coll = '#D55E00'

# ========== DERIVED ==========
q_star = (a - c) / (3 * b)      # 40
q_mono = (a - c) / (2 * b)      # 60

q_range = np.linspace(0, (a - c) / b, 200)

# Best response functions
br1 = np.maximum(0, (a - c) / (2 * b) - q_range / 2)
br2 = np.maximum(0, (a - c) / (2 * b) - q_range / 2)

fig, ax = plt.subplots(figsize=(7, 5.5))

ax.plot(q_range, br1, '-', color=color1, linewidth=2.5, label="Firm 1's BR: $q_1^*(q_2)$")
ax.plot(br2, q_range, '-', color=color2, linewidth=2.5, label="Firm 2's BR: $q_2^*(q_1)$")

# Cournot equilibrium
ax.plot(q_star, q_star, 'o', color=color_eq, markersize=10, zorder=5)
ax.annotate(f'Cournot NE\n$q^* = {q_star:.0f}$',
            xy=(q_star, q_star), xytext=(q_star + 15, q_star + 10),
            fontsize=12, ha='center',
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5))

# Collusion point
ax.plot(q_mono / 2, q_mono / 2, 's', color=color_coll, markersize=8, zorder=5)
ax.annotate('Collusion\n($q = 30$ each)',
            xy=(q_mono / 2, q_mono / 2), xytext=(q_mono / 2 + 18, q_mono / 2 - 8),
            fontsize=11, ha='center', color=color_coll,
            arrowprops=dict(arrowstyle='->', color=color_coll, lw=1.2))

# Deviation point
q_dev = 45
ax.plot(q_dev, q_mono / 2, 'D', color='#CC79A7', markersize=7, zorder=5)
ax.annotate('Firm 1 deviates\n($q_1 = 45, q_2 = 30$)',
            xy=(q_dev, q_mono / 2), xytext=(q_dev + 12, q_mono / 2 + 14),
            fontsize=10, ha='center', color='#CC79A7',
            arrowprops=dict(arrowstyle='->', color='#CC79A7', lw=1.2))

ax.set_xlabel('$q_1$', fontsize=14)
ax.set_ylabel('$q_2$', fontsize=14)
ax.set_xlim(0, 130)
ax.set_ylim(0, 130)
ax.legend(fontsize=11, loc='upper right')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
