```python
#| echo: false
#| fig-align: center

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Polygon

# ========== PARAMETERS ==========
demand_intercept = 50      # P-intercept of demand  (Q = 5000 - 100P => P = 50 - Q/100)
demand_slope     = -1/100  # dP/dQ, demand
supply_slope     = 1/150   # dP/dQ, domestic supply (Q = 150P => P = Q/150)

P_world  = 10.0   # free-trade world price
P_tariff = 15.0   # world price + $5 tariff
P_eq     = 20.0   # no-trade domestic equilibrium price

# ========== COLORS (Okabe-Ito palette) ==========
C_BLUE   = '#0072B2'   # demand curve
C_ORANGE = '#E69F00'   # supply curve / region A
C_GREEN  = '#009E73'   # world price line / region B
C_RED    = '#D55E00'   # tariff price line / DWL regions C & D
C_GRAY   = '#999999'   # dotted drop lines

# ========== DERIVED QUANTITIES ==========
Q_S_world  = P_world  / supply_slope                        # 1,500
Q_D_world  = (demand_intercept - P_world)  / (-demand_slope) # 4,000
Q_S_tariff = P_tariff / supply_slope                        # 2,250
Q_D_tariff = (demand_intercept - P_tariff) / (-demand_slope) # 3,500

# ========== FIGURE ==========
fig, ax = plt.subplots(figsize=(8, 6))

Q_range  = np.linspace(0, 5200, 600)
P_demand = demand_intercept + demand_slope * Q_range
P_supply = supply_slope * Q_range

# --- Supply and demand curves ---
ax.plot(Q_range[P_demand >= 0],
        P_demand[P_demand >= 0],
        color=C_BLUE,   linewidth=2.2, label='Demand')
ax.plot(Q_range[P_supply <= 55],
        P_supply[P_supply <= 55],
        color=C_ORANGE, linewidth=2.2, label='Domestic supply')

# --- Horizontal price lines ---
ax.axhline(P_world,  color=C_GRAY, linewidth=1.6,
           linestyle='--')
ax.axhline(P_tariff, color=C_GRAY,   linewidth=1.6,
           linestyle='--')

# --- Shaded welfare regions ---
regions = [
    # (vertices, facecolor, legend label)

    ([(Q_S_world,  P_world), (Q_S_tariff, P_world), (Q_S_tariff, P_tariff)],
     C_RED, 'DWL (A + C)'),

    ([(Q_S_tariff, P_world),  (Q_D_tariff, P_world),
      (Q_D_tariff, P_tariff), (Q_S_tariff, P_tariff)],
     C_GREEN, 'Tariff revenue (B)'),     

    ([(Q_D_tariff, P_world), (Q_D_world,  P_world), (Q_D_tariff, P_tariff)],
     C_RED, None),

    ([(0,  P_world),  (Q_S_tariff, P_world),
      (Q_S_tariff, P_tariff), (0,  P_tariff)],
     C_ORANGE, 'CS → PS transfer (D)'),     
]
for verts, color, label in regions:
    ax.add_patch(Polygon(verts, closed=True,
                         facecolor=color, alpha=0.45,
                         edgecolor='none', label=label))

# --- Vertical drop lines and point markers ---
key_pts = [(Q_S_world, P_world), (Q_S_tariff, P_tariff),
           (Q_D_tariff, P_tariff), (Q_D_world,  P_world)]
for Q_val, P_val in key_pts:
    ax.plot([Q_val, Q_val], [0, P_val],
            color=C_GRAY, linewidth=1.0, linestyle=':', zorder=1)
    ax.plot(Q_val, P_val, 'o', color=C_GRAY, markersize=5, zorder=5)

# --- Region letter labels ---
lk = dict(fontsize=11, fontweight='bold', ha='center', va='center', color='white')
ax.text((Q_S_tariff) / 2 , (P_world + P_tariff) / 2.1, 'D', **lk)
ax.text((Q_S_tariff + Q_D_tariff) / 2,  (P_world + P_tariff) / 2.1, 'B', **lk)
ax.text(Q_S_world  + (Q_S_tariff - Q_S_world)  * 0.65, (P_world + P_tariff) / 2.1, 'C', **lk)
ax.text(Q_D_tariff + (Q_D_world  - Q_D_tariff) * 0.35, (P_world + P_tariff) / 2.1, 'A', **lk)

# --- Axes formatting ---
ax.set_xlabel('Quantity', fontsize=12)
ax.set_ylabel('Price', fontsize=12)
ax.set_xlim(0, 5100)
ax.set_ylim(0, 52)

ax.set_xticks([Q_S_world, Q_S_tariff, Q_D_tariff, Q_D_world])
ax.set_xticklabels([
    f'$Q_S^{{FT}}$\n({int(Q_S_world):,})',
    f'$Q_S^{{T}}$\n({int(Q_S_tariff):,})',
    f'$Q_D^{{T}}$\n({int(Q_D_tariff):,})',
    f'$Q_D^{{FT}}$\n({int(Q_D_world):,})',
], fontsize=9)
ax.set_yticks([0, P_world, P_tariff, P_eq])
ax.set_yticklabels(['0', '$P_W = 10$', r'$P_W + t = 15$', '$P^* = 20$'], fontsize=9)

ax.legend(loc='upper right', fontsize=9, framealpha=0.85)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
